### **Agentic AI**

**Agentic AI is a type of an AI that can take up a task/goal from a user and then work toward completing it on its own, with minimal human intervention/guidance. It plans, takes action, adapts to changes and seek help only necessary**

### **Generative AI**

**Generative AI (GenAI) is a type of artificial intelligence that creates new content—like text, images, audio, video, and code—by learning patterns from existing data. Instead of just sorting or analyzing data, GenAI builds unique, original responses based on simple user prompts**.

#### **AI Agents** is someone who performs the specific task, when **LLMs** perform the task with any human intervention and can achieve the goals with minimal human intervention.

* *An AI agent is an autonomous system designed to achieve a specific goal by perceiving its environment, making decisions, and taking multi-step actions using external tools.Unlike traditional software or static chatbots that only reply to direct prompts, an AI agent takes an overarching goal, breaks it down into subtasks, and executes them independently*.

    - **Core Components**:
    * **Brain**: Decides what to do next.
    * **Planning**: Devises steps and reflects on mistakes to fix them.
    * **Memory**: Remembers past conversations and prior actions.
    * **Tools**: Uses APIs, web search, or code execution to act in the real world.
    
    - **The Role of the LLM (Large Language Model)**
    * **The Brain**: The Large Language Model acts as the central intelligence engine for the agent.
    * **Understanding Intent**: It parses natural language instructions from users to map out what the final objective means.
    * **Reasoning and Planning**: It figures out the logical sequence of steps needed to accomplish a task and decides which tool to call next.
    * **Reflection**: It analyzes the output from previous tool actions to evaluate if it succeeded or if it needs to try a different approach.

#### Agentic AI

* **An AI agent is an individual program built for a specific task, whereas agentic AI describes a coordinated system of multiple agents working together toward a broader goal**.

    - **Key Differences**:
    * **Scope and Focus**:
        * **AI Agents**: Focus on a single, narrow, and repeatable job (like answering one support question or booking a flight).
        * **Agentic AI**: Focuses on high-level outcomes by managing complex, multi-step workflows across systems.
        
    * **Autonomy**:
        * **AI Agents**: Operate reactively within fixed rules and require explicit prompts or triggers.
        * **Agentic AI**: Proactively sets sub-goals, plans actions, uses tools, and adapts when obstacles appear.
        
    * **Architecture**:
        * **AI Agents**: Act like a single specialized worker.
        * **Agentic AI**: Acts like a manager or orchestrator that delegates work to multiple specialized agents

#### Langchain

**LangChain is an open-source software framework that makes it easier to build applications using large language models (LLMs) by connecting them to external data, tools, and multi-step workflows**.

* **Key Components**:
    * **Models and Prompts**: Provides standard interfaces to connect with various LLM providers (like OpenAI or Anthropic) and manage prompt templates.
    * **Chains**: Combines LLMs with other functions in a sequence, allowing the output of one step to become the input for the next.
    * **Memory**: Keeps track of conversation history or summaries so that models can remember past interactions.
    * **Agents**: Uses an LLM as a reasoning engine to decide which external tools or APIs to call to accomplish a task.
    * **Retrieval (RAG)**: Connects models to external documents, databases, and vector stores to pull in relevant facts during runtime

* Core Components of LangChain
    
    - **Models**: Standardized interfaces to interact with various AI models, including LLMs, chat models, and embedding models (which convert text into numerical vectors). 
    - **Prompts**: Templates and management tools designed to handle dynamic input variables, formatting, and consistent user instructions. 
    - **Chains**: Sequences of calls (or pipelines) combining models, prompts, and data processing steps to execute multi-step workflows. 
    - **Memory**: Systems that preserve context and past message history across conversations. 
    - **Indexes / Retrievers**: Tools for document loading, splitting, vector storage (like Chroma or FAISS), and retrieval-augmented generation (RAG). 
    - **Agents**: LLM-driven orchestrators that reason through problems, select appropriate external tools, and execute iterative loops until a task completes. 
    - **Tools**: External utilities or APIs that agents can invoke to perform live web searches, computations, or database lookups
