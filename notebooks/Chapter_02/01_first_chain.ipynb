### Chains in LangChain

**LangChain chains combine prompts, models, and extra tools in a sequence to run step-by-step workflows for artificial intelligence applications**.

* **What is a Chain**?
    * A chain links multiple tasks together so the output of one step becomes the input for the next step.
    * Modern versions of LangChain use the pipe operator (|) and Runnables instead of the older LLMChain class.
    * This setup separates your app code from any single model provider.
    
* For a beginner-friendly overview of how LangChain links steps together in a sequence:
    * **Core Types of Chains**
        - **LLMChain (Prompt | Model)**: The basic unit that takes an input, formats a prompt, and sends it to a language model.
        - **Sequential Chains**: Links multiple chains in a line, passing the result from one stage directly forward.
        - **Router Chain**: Uses conditional branching logic to decide which sub-chain should handle a specific input.
        - **Retrieval QA**: Connects a language model to a vector database to answer questions based on external documents

```python
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from dotenv import load_dotenv

## This function will load all the variable from .env file and will make them available in
## os.environ directory (env_variablea)
load_dotenv()

import os 

if os.environ.get("OPENAI_API_KEY"):
    print("✅ OPENAI_API_KEY Exists.")
else:
    raise ValueError("❌ OPENAI_API_KEY Not Found...")
```

### First Chain

```python
# Task 1: Prompt
prompt_template = ChatPromptTemplate.from_messages([
    ('system', 'You are helpful assistant'),
    ('human', "{input}")
])
```

```python
# Task 2: LLM
llm_openai = ChatOpenAI(model='gpt-5-mini',
                        temperature=0)
```

```python
# Task 3; Output parser
from langchain_core.output_parsers import StrOutputParser

str_parser = StrOutputParser()
```

### Manual Invoking

```python
template = prompt_template.invoke({'input': 'What is the Capital of Sweden'})
result = llm_openai.invoke(template)
result.content
```

### Chain Invocation

```python
## First Way: with Pipe Symbol
chain = prompt_template | llm_openai | str_parser

chain.invoke({'input': 'What is the Capital of Sweeden'})
```

```python
## Second way: RunableSequence
from langchain_core.runnables import RunnableSequence

chain_run = RunnableSequence(
    prompt_template, llm_openai, str_parser
)

chain_run.invoke({'input': 'What is the Capital of Sweeden'})
```
