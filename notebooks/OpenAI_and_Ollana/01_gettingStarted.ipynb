```python
import os
from dotenv import load_dotenv

## This function will load all the variable from .env file and will make them available
## os.environ directory (env_variablea)
load_dotenv()

os.environ['OPENAI_API_KEY'] = os.getenv("OPENAI_API_KEY")

## for langsmith tracking
os.environ['LANGCHAIN_API_KEY'] = os.getenv("LANGCHAIN_API_KEY")
os.environ['LANGCHAIN_TRACING_V2'] = "true"
os.environ['LANGCHAIN_PROJECT'] = os.getenv("LANGCHAIN_PROJECT")
```

```python
from langchain_openai import ChatOpenAI

openai_llm = ChatOpenAI(model='gpt-4o')
print(openai_llm)
```

```python
### Input and get response
result = openai_llm.invoke("What is Generative AI?")
```

### ChatPrompt Template

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        ('system', "You are an expert AI Engineer, provide me the answer based on your experience"),
        ('user', "{input}")
    ]
)

prompt
```

```python
## Chain
chain = prompt | openai_llm
response = chain.invoke({'input':"Tell me about langsmith"})
print(response)
```
