### Langchain messages 

```python
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv

## This function will load all the variable from .env file and will make them available in
## os.environ directory (env_variablea)
load_dotenv()

import os 

if os.environ.get("OPENAI_API_KEY"):
    print("✅ OPENAI_API_KEY Exists.")
else:
    raise ValueError("❌ OPENAI_API_KEY Not Found...")
```

```python
llm_openai = ChatOpenAI(model="gpt-5-mini",
                    temperature=0)

my_messages = [
    SystemMessage(content="Explain in a funny way"),  ## setting the tone
    HumanMessage(content="hey, How are you doing?")
]

llm_openai.invoke(my_messages).content
```

### **Prompts**

**Promts are the messages that are sent to the LLMs; Promts are more use friendly than messages**.

```python
my_messages = [
    SystemMessage(content='Explain in a creative way.'),
    HumanMessage(content="tell me a fun fact about honey")
]

llm_openai.invoke(my_messages).content
```

```python
from langchain_core.prompts import PromptTemplate

user_input = input("Enter a topic for the fact: ") 

dynamic_prompt = PromptTemplate.from_template("Write a fun fact about {topic}")

read_prompt = dynamic_prompt.invoke({'topic': user_input})

llm_openai.invoke(read_prompt).content
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt_template = ChatPromptTemplate.from_messages([
    ('system', "you are a {tone} assistant"),
    ('human', "tell me a fun fact about {topic}, return me the result in key value pair format with  \
     key as 'fact' and value as 'fun fact'")
])

## Multiple input
input_prompt = input("Enter a topic for the fact: ")
tone_prompt = input("Enter the tone you wanna set for the assistant:")

# multiple input prompts mapping
ready_prompt = prompt_template.invoke({'topic': input_prompt, 'tone': tone_prompt})
print(ready_prompt.messages)

llm_openai.invoke(ready_prompt).content
```
