### Stream

**Most model can stream their comtent while it is being generated. By displaying output progressively, streaming significantly improves the user experience, particularly for longer responses.Calling `Stream()` returns an iterator tat yields the output chunks as they are produced.**

```python
model.invoke('write me a 200 word paragraph on Bhagat Singh')
```

```python
for chunk in model.stream("write me a 200 word paragraph on Artificial Intelligence"):
    if chunk.content:
        print(chunk.content, end=" ", flush=True)
```

```python
for chunk in model.stream("why do parrots have colorful feathers"):
    print(chunk.content, end=" ", flush=True)
```

### Batch

**Batching a collection of independent request to a model can significantly improve the performance and reduced costs as the processing can be done in parallel.**

```python
responses = model.batch(
    ["why do parrots have colorful feathers",
    "How do airplane flies",
    "What is Quantum computing",
    "Humbolt Enigma"],
    config={
        'max_concurrency':3  # limit to max 3 parallel calls
    })

for response in responses:
    print(response.content)
```

### Tools


Models can request to call tools that perform tasks such as fetching data from a database, searching the web, or running code. Tools are pairings of:

1. A schema, including the name of the tool, a description, and/or argument definitions (often a JSON schema)
2. A function or coroutine to execute.

```python
from langchain.tools import tool

@tool
def get_weather(location:str) -> str:
    '''Get weather at a location'''
    return f"It is Sunny in {location}"

model_with_tools = model.bind_tools([get_weather])
```

```python
response = model_with_tools.invoke("What's the weather like in Singapore")
print(response)

for tool_call in response.tool_calls:
    ## View tool call made by model
    print(f'Tool: {tool_call['name']}')
    print(f'Tool: {tool_call['args']}')
```

### Tool Execution Loop

```python
# Step 1: Model generates tool calls
messages = [
    {
        "role": "user",
        "content": "What's the weather in Boston?"
    }
]

ai_msg = model_with_tools.invoke(messages)
messages.append(ai_msg)

# Step 2: Execute tools and collect results
for tool_call in ai_msg.tool_calls:
    tool_result = get_weather.invoke(tool_call)
    messages.append(tool_result)

# Step 3: Pass results back to model for final response
final_response = model_with_tools.invoke(messages)
print(final_response.text)
```
