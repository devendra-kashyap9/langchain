### Web Based Loader

```python
import bs4
from langchain_community.document_loaders import  WebBaseLoader

web_loader = WebBaseLoader(web_path=("https://sebastianraschka.com/faq/docs/rope-vs-absolute-positional-embeddings.html",),
                           bs_kwargs=dict(parse_only=bs4.SoupStrainer(
                               class_=("post-title", "post-content", "post-header")
                           ))
                           )

web_loader.load()
```

### Arxiv Loader

```python
import arxiv

# Force HTTPS endpoint to prevent HTTP 301 redirect errors
arxiv.Result.ROOT_URL = "https://export.arxiv.org/api/"

from langchain_community.document_loaders import ArxivLoader

docs = ArxivLoader(query="1706.03762", load_max_docs=2).load()
print(f"Loaded {len(docs)} document(s)")
```

```python
from langchain_community.document_loaders import PyPDFLoader

# Direct HTTPS PDF download URL for arXiv
arxiv_id = "1706.03762"
loader = PyPDFLoader(f"https://arxiv.org/pdf/{arxiv_id}.pdf")

docs = loader.load()
print(f"Loaded {len(docs)} pages")
print("Title/First page preview:", docs[0].page_content[:300])
```

```python
from langchain_community.document_loaders import ArxivLoader

loader = ArxivLoader(
    query="reasoning"
)

docs = loader.get_summaries_as_docs()
print(docs[0].page_content[:100])
print(docs[0].metadata)
```
