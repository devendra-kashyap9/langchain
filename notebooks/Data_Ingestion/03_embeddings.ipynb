### Embeddings

**Embeddings are lists of numbers that represent the meaning of complex data like words, sentences, or images so computers can understand their relationships**. 

* **What Are Embeddings?**
    - **Dense vectors:** Large groups of numbers (often hundreds or thousands) that store data features.
    - **Semantic space:** A virtual map where items with similar meanings sit close to each other.
    - **Computer memory:** A way to translate human concepts into math that machine learning models can read. 
    
* *Why Do We Need Them?*
    - Computers cannot read raw text, images, or audio directly. They need numbers.
    - Older methods like one-hot encoding made massive, empty lists that wasted memory and missed connections.
    - Embeddings shrink data size while keeping the true meaning intact

### OpenAIEmbeddings

```python
from langchain_openai import  OpenAIEmbeddings

openai_embeddings = OpenAIEmbeddings(model="text-embedding-3-large")
openai_embeddings
```

```python
text = "This is an openAI embeddings test"
embeddings = openai_embeddings.embed_query(text)
embeddings
```

```python
len(embeddings)
```

```python
openai_embeddings_1024 = OpenAIEmbeddings(model="text-embedding-3-large",
                                     dimensions=1024)
openai_embeddings_1024

text = "This is an openAI embeddings test"
embeddings = openai_embeddings_1024.embed_query(text)
len(embeddings)
```

```python
from langchain_community.document_loaders import TextLoader

txt_loader = TextLoader('speech.txt')
docs = txt_loader.load()
docs
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=100,
                                               chunk_overlap=50
                                               )
final_documents = text_splitter.split_documents(docs)
final_documents
```

### VectorEmbedding and VectorStore

```python
from langchain_community.vectorstores import Chroma

db = Chroma.from_documents(final_documents, 
                           openai_embeddings_1024)

db
```

```python
query = "We must educate ourselves, we must awaken our youth"
retrieved_result = db.similarity_search(query)
print(retrieved_result)
```

### OllamaEmbeddings

**Ollama is a free, open-source platform that lets you download and run large language models locally on your own computer, Ollama supports vector embeddings to turn text into number arrays for semantic search and Retrieval-Augmented Generation (RAG) applications**.

```python
from langchain_community.embeddings import OllamaEmbeddings

ollama_embeddings = OllamaEmbeddings(model="gemma:2b")  # default model is llama2
ollama_embeddings
```

### HuggingFaceEmbeddings


**Hugging Face Sentence Transformers is an open-source Python framework designed for computing dense vector representations (embeddings) for sentences, paragraphs, and images.** 

* **Key Features**
    - **Semantic Embeddings:** Maps text or images into a dense vector space where similar concepts are positioned close together.
    - **Multiple Modalities:** Supports computing embeddings for text, images, audio, and video.
    - **Cross-Encoders & Rerankers:** Allows calculation of precise similarity scores by comparing pairs of texts directly through reranker models.
    - **Local Execution:** Runs locally on your machine without requiring an external API key

```python
import os
from dotenv import load_dotenv

load_dotenv()
os.environ['HF_TOKEN'] = os.getenv('HF_TOKEN')
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

huggingface_embeddings = HuggingFaceEmbeddings(
    model="all-MiniLM-L6-v2"
)
```

```python
text = "This is a HuggingFace embeddings test"
query_result = huggingface_embeddings.embed_query(text)
query_result
```
